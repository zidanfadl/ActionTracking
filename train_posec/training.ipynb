## Training via bash

```python
import os
import re

base_dir = '../work_dirs'
base_name = 'ciis_21_v'

# List matching directories
versions = [
    int(m.group(1)) for name in os.listdir(base_dir)
    if (m := re.match(fr'{re.escape(base_name)}(\d+)', name)) and os.path.isdir(os.path.join(base_dir, name))
]
next_version = max(versions, default=0) + 1

# Final path
work_dir = os.path.join(base_dir, f'{base_name}{next_version}')
print(work_dir)
```

```python
!CUDA_VISIBLE_DEVICES=1
```

```python
!python ../mmaction2/tools/train.py ../configs/skeleton/posec3d/ciis_10_mod.py \
    --work-dir $work_dir \
    --seed 0 
```
