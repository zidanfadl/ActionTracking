```python
import os
import math
from PIL import Image

def make_grid_photo(folder_path, output_path='grid.jpg', image_size=(1280, 720), padding=10, bg_color=(255, 255, 255)):
    extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')
    image_files = sorted([
        os.path.join(folder_path, file) 
        for file in os.listdir(folder_path) 
        if file.lower().endswith(extensions)
    ])

    photo_count = len(image_files)
    if photo_count == 0:
        print("No images found in folder.")
        return

    cols = math.ceil(math.sqrt(photo_count))
    rows = math.ceil(photo_count / cols)

    grid_width = cols * image_size[0] + (cols + 1) * padding
    grid_height = rows * image_size[1] + (rows + 1) * padding

    grid_img = Image.new('RGB', (grid_width, grid_height), bg_color)

    for idx, image_path in enumerate(image_files):
        img = Image.open(image_path).convert("RGB")
        img = img.resize(image_size)

        row = idx // cols
        col = idx % cols

        x = padding + col * (image_size[0] + padding)
        y = padding + row * (image_size[1] + padding)

        grid_img.paste(img, (x, y))

    grid_img.save(output_path)
    print(f"Grid image saved to: {output_path}")

# Example
make_grid_photo('assets/video/data1')
```
