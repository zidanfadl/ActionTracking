```python
#modified json log structure

import json
from collections import OrderedDict

def add_mode_to_json(input_file_path, output_file_path):
    with open(input_file_path, 'r') as file:
        lines = file.readlines()

    updated_lines = []
    for line in lines:
        data = json.loads(line)
        first_key = list(data.keys())[0]
        mode = "train" if first_key == "lr" else "val" if first_key == "acc/top1" else None
        if mode:
            new_data = OrderedDict([("mode", mode)])
            new_data.update(data)
            if mode == "val":
                step = data.get("step")
                new_data["iter"] = step
                new_data["epoch"] = step
            updated_lines.append(json.dumps(new_data))

    with open(output_file_path, 'w') as file:
        file.write("\n".join(updated_lines))

# Replace 'input_file_path.json' and 'output_file_path.json' with the actual paths to your JSON files
add_mode_to_json(
    'work_dirs/ciis_21_v4/ciis_21_v4.json',
    'work_dirs/ciis_21_v4/ciis_21_v4_modified.json'
)
```

```python
def concatenate_json_lines_files(file_paths: list, output_file: str) -> None:
    """
    Concatenates multiple JSON Lines files while preserving the original format
    (one JSON object per line, no enclosing array brackets)
    
    Args:
        file_paths: List of paths to JSON Lines files to concatenate
        output_file: Path to save the concatenated JSON file
    """
    processed_files = 0
    total_items = 0
    
    try:
        with open(output_file, 'w', encoding='utf-8') as out_f:
            for file_path in file_paths:
                try:
                    # Check if file exists
                    if not os.path.exists(file_path):
                        display(Markdown(f"⚠️ **Warning**: File not found - `{file_path}`"))
                        continue
                        
                    # Read and process each line
                    file_items = 0
                    with open(file_path, 'r', encoding='utf-8') as in_f:
                        for line in in_f:
                            line = line.strip()
                            if line:  # Skip empty lines
                                # Validate JSON (but keep original format)
                                try:
                                    _ = json.loads(line)  # Validate JSON syntax
                                    out_f.write(line + '\n')
                                    file_items += 1
                                except json.JSONDecodeError as e:
                                    display(Markdown(f"⚠️ **Warning**: Invalid JSON line in `{file_path}` - {str(e)}"))
                    
                    if file_items > 0:
                        processed_files += 1
                        total_items += file_items
                        display(Markdown(f"✅ Processed {file_items} items from `{file_path}`"))
                    else:
                        display(Markdown(f"⚠️ **Warning**: No valid JSON objects found in `{file_path}`"))
                            
                except Exception as e:
                    display(Markdown(f"❌ **Error** processing `{file_path}` - `{str(e)}`"))
        
        display(Markdown(f"## 🎉 Successfully concatenated {total_items} items from {processed_files} files to `{output_file}`"))
        display(Markdown(f"## Output format matches original (one JSON object per line)"))
    
    except Exception as e:
        display(Markdown(f"❌ **Error** writing output file - `{str(e)}`"))
```

```python
def interactive_concatenation():
    display(Markdown("## 📂 JSON Lines File Concatenator (Interactive Mode)"))
    
    input_files = []
    while True:
        path = input("Enter JSON file path (or 'done' to finish): ").strip()
        if path.lower() == 'done':
            break
        input_files.append(path)
    
    if not input_files:
        display(Markdown("❌ No files were provided"))
        return
    
    output_file = input("Enter output file path [default: combined_output.json]: ").strip()
    output_file = output_file if output_file else 'combined_output.json'
    
    concatenate_json_lines_files(input_files, output_file)
```
